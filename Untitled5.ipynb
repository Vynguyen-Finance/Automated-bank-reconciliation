```python
import pandas as pd

# 1. Tạo dữ liệu Sổ cái (GL)
gl_data = {
    'Date': ['2026-06-01', '2026-06-02', '2026-06-03', '2026-06-04'],
    'Voucher_No': ['PC001', 'PT002', 'PC003', 'PC004'],
    'Description': ['Chi tien mua van phong pham', 'Khach hang A thanh toan', 'Tra tien nha cung cap B', 'Ke toan ghi trung khoan nay'],
    'Amount': [-1500000, 25000000, -8500000, -3000000]
}
df_gl = pd.DataFrame(gl_data)
df_gl.to_excel('GL_Data.xlsx', index=False)

# 2. Tạo dữ liệu Ngân hàng (Bank)
bank_data = {
    'Date': ['2026-06-01', '2026-06-04', '2026-06-03', '2026-06-05'],
    'Transaction_ID': ['FT26012', 'FT26015', 'FT26016', 'FT26019'],
    'Description': ['CT MUA VPP', 'KH A CHUYEN KHOAN', 'TRA TIEN NCC B', 'PHI QUAN LY TAI KHOAN'],
    'Amount': [-1500000, 25000000, -8500000, -55000]
}
df_bank = pd.DataFrame(bank_data)
df_bank.to_excel('Bank_Data.xlsx', index=False)

print("Đã tạo xong 2 file dữ liệu mẫu: GL_Data.xlsx và Bank_Data.xlsx!")
```

```python
import pandas as pd
import numpy as np

def bank_reconciliation(gl_path, bank_path, output_path):
    # 1. ĐỌC DỮ LIỆU TỪ FILE EXCEL ĐÃ LƯU TRÊN COLAB
    df_gl = pd.read_excel(gl_path)
    df_bank = pd.read_excel(bank_path)

    # Chuẩn hóa định dạng Ngày tháng để Python tính toán khoảng cách ngày
    df_gl['Date'] = pd.to_datetime(df_gl['Date'])
    df_bank['Date'] = pd.to_datetime(df_bank['Date'])

    # Tạo thêm cột Match_ID để đánh dấu các dòng đã được khớp
    df_gl['Match_ID'] = np.nan
    df_bank['Match_ID'] = np.nan

    match_counter = 1

    # 2. VÒNG LẶP ĐỐI CHIẾU THÔNG MINH
    for idx_gl, row_gl in df_gl.iterrows():
        # Điều kiện khớp: Trùng số tiền VÀ chênh lệch không quá 3 ngày VÀ dòng đó chưa bị khớp trước đó
        condition = (
            (df_bank['Amount'] == row_gl['Amount']) &
            (df_bank['Match_ID'].isna()) &
            (abs((df_bank['Date'] - row_gl['Date']).dt.days) <= 3)
        )

        matched_bank = df_bank[condition]

        if not matched_bank.empty:
            idx_bank = matched_bank.index[0] # Lấy dòng đầu tiên tìm thấy

            # Đánh dấu mã MATCH giống nhau cho cả 2 bên
            df_gl.at[idx_gl, 'Match_ID'] = f"MATCH_{match_counter}"
            df_bank.at[idx_bank, 'Match_ID'] = f"MATCH_{match_counter}"
            match_counter += 1

    # 3. PHÂN LOẠI CÁC DÒNG CHƯA KHỚP (CHÊNH LỆCH) VÀ ĐÃ KHỚP
    gl_unmatched = df_gl[df_gl['Match_ID'].isna()]
    bank_unmatched = df_bank[df_bank['Match_ID'].isna()]
    gl_matched = df_gl[df_gl['Match_ID'].notna()]

    # 4. XUẤT KẾT QUẢ RA MỘT FILE EXCEL MỚI NHIỀU SHEETS
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        gl_unmatched.to_excel(writer, sheet_name='Sổ cái thiếu ở Bank', index=False)
        bank_unmatched.to_excel(writer, sheet_name='Bank thiếu ở Sổ cái', index=False)
        gl_matched.to_excel(writer, sheet_name='Đã khớp (Sổ cái)', index=False)

    print(f"🎉 Đối chiếu hoàn tất thành công!")
    print(f"👉 File kết quả đã được lưu tại bộ nhớ Colab với tên: {output_path}")

# KÍCH HOẠT CHẠY THỰC TẾ
bank_reconciliation('GL_Data.xlsx', 'Bank_Data.xlsx', 'Reconciliation_Report.xlsx')
```

```python
import pandas as pd
import numpy as np

def bank_reconciliation_v2(gl_path, bank_path, output_path):
    # 1. ĐỌC DỮ LIỆU
    df_gl = pd.read_excel(gl_path)
    df_bank = pd.read_excel(bank_path)

    df_gl['Date'] = pd.to_datetime(df_gl['Date'])
    df_bank['Date'] = pd.to_datetime(df_bank['Date'])

    df_gl['Match_ID'] = np.nan
    df_bank['Match_ID'] = np.nan

    match_counter = 1

    # 2. VÒNG LẶP ĐỐI CHIẾU
    for idx_gl, row_gl in df_gl.iterrows():
        condition = (
            (df_bank['Amount'] == row_gl['Amount']) &
            (df_bank['Match_ID'].isna()) &
            (abs((df_bank['Date'] - row_gl['Date']).dt.days) <= 3)
        )
        matched_bank = df_bank[condition]
        if not matched_bank.empty:
            idx_bank = matched_bank.index[0]
            df_gl.at[idx_gl, 'Match_ID'] = f"MATCH_{match_counter}"
            df_bank.at[idx_bank, 'Match_ID'] = f"MATCH_{match_counter}"
            match_counter += 1

    # 3. PHÂN LOẠI CÁC DÒNG CHƯA KHỚP VÀ ĐÃ KHỚP
    gl_unmatched = df_gl[df_gl['Match_ID'].isna()]
    bank_unmatched = df_bank[df_bank['Match_ID'].isna()]
    gl_matched = df_gl[df_gl['Match_ID'].notna()]

    # 4. TÍNH TOÁN CÁC GIÁ TRỊ TỔNG ĐỂ LÀM TÓM TẮT (Đưa vào trong hàm)
    total_gl_unmatched = gl_unmatched['Amount'].sum()
    total_bank_unmatched = bank_unmatched['Amount'].sum()

    summary_data = {
        'Chỉ tiêu': [
            '1. Tổng số tiền chưa khớp từ phía Sổ Cái (Cần điều chỉnh)',
            '2. Tổng số tiền chưa khớp từ phía Ngân Hàng (Cần kiểm tra)',
            'Kết luận'
        ],
        'Số tiền (VND)': [
            total_gl_unmatched,
            total_bank_unmatched,
            'Cần rà soát chênh lệch' if total_gl_unmatched != 0 or total_bank_unmatched != 0 else 'Khớp hoàn toàn'
        ]
    }
    df_summary = pd.DataFrame(summary_data)

    # 5. XUẤT TẤT CẢ VÀO 1 FILE EXCEL DUY NHẤT (Nhiều sheets)
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        df_summary.to_excel(writer, sheet_name='Tóm tắt (Summary)', index=False)
        gl_unmatched.to_excel(writer, sheet_name='Sổ cái thiếu ở Bank', index=False)
        bank_unmatched.to_excel(writer, sheet_name='Bank thiếu ở Sổ cái', index=False)
        gl_matched.to_excel(writer, sheet_name='Đã khớp (Sổ cái)', index=False)

    print(f"🎉 Xuất báo cáo thành công! Hãy tải file {output_path} về.")

# KÍCH HOẠT CHẠY THỰC TẾ
bank_reconciliation_v2('GL_Data.xlsx', 'Bank_Data.xlsx', 'Reconciliation_Report_Full.xlsx')
```

```python
import pandas as pd
import numpy as np
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

def bank_reconciliation_fdi(gl_path, bank_path, output_path):
    # 1. ĐỌC DỮ LIỆU
    df_gl = pd.read_excel(gl_path)
    df_bank = pd.read_excel(bank_path)

    df_gl['Date'] = pd.to_datetime(df_gl['Date'])
    df_bank['Date'] = pd.to_datetime(df_bank['Date'])

    df_gl['Match_ID'] = np.nan
    df_bank['Match_ID'] = np.nan

    match_counter = 1

    # 2. VÒNG LẶP ĐỐI CHIẾU
    for idx_gl, row_gl in df_gl.iterrows():
        condition = (
            (df_bank['Amount'] == row_gl['Amount']) &
            (df_bank['Match_ID'].isna()) &
            (abs((df_bank['Date'] - row_gl['Date']).dt.days) <= 3)
        )
        matched_bank = df_bank[condition]
        if not matched_bank.empty:
            idx_bank = matched_bank.index[0]
            df_gl.at[idx_gl, 'Match_ID'] = f"MATCH_{match_counter}"
            df_bank.at[idx_bank, 'Match_ID'] = f"MATCH_{match_counter}"
            match_counter += 1

    # 3. PHÂN LOẠI (Chuyển tên sheet và cột sang tiếng Anh)
    gl_unmatched = df_gl[df_gl['Match_ID'].isna()].drop(columns=['Match_ID'])
    bank_unmatched = df_bank[df_bank['Match_ID'].isna()].drop(columns=['Match_ID'])
    matched_data = df_gl[df_gl['Match_ID'].notna()]

    # 4. TẠO BẢNG TÓM TẮT TIẾNG ANH (SUMMARY ACCOUNTING FORM)
    total_gl_unmatched = gl_unmatched['Amount'].sum()
    total_bank_unmatched = bank_unmatched['Amount'].sum()

    summary_data = {
        'Financial Items': [
            '1. Total Unmatched Items from General Ledger (GL)',
            '2. Total Unmatched Items from Bank Statement',
            'Status / Conclusion'
        ],
        'Amount (VND)': [
            total_gl_unmatched,
            total_bank_unmatched,
            'Pending Review' if total_gl_unmatched != 0 or total_bank_unmatched != 0 else 'Fully Reconciled'
        ]
    }
    df_summary = pd.DataFrame(summary_data)

    # 5. XUẤT RA EXCEL VÀ TỰ ĐỘNG FORMAT ĐẸP
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        df_summary.to_excel(writer, sheet_name='Summary', index=False)
        gl_unmatched.to_excel(writer, sheet_name='GL_Unmatched', index=False)
        bank_unmatched.to_excel(writer, sheet_name='Bank_Unmatched', index=False)
        matched_data.to_excel(writer, sheet_name='Matched_Items', index=False)

        # --- BẮT ĐẦU QUY TRÌNH FORMAT ĐẸP (STYLING) ---
        workbook = writer.book

        # Định nghĩa các Style chuẩn Corporate
        font_family = "Segoe UI"
        header_fill = PatternFill(start_color="1F4E78", end_color="1F4E78", fill_type="solid") # Màu xanh Navy đậm (chuẩn FDI)
        header_font = Font(name=font_family, size=11, bold=True, color="FFFFFF")
        regular_font = Font(name=font_family, size=10)
        bold_font = Font(name=font_family, size=10, bold=True)

        thin_border_side = Side(border_style="thin", color="D9D9D9")
        thin_border = Border(left=thin_border_side, right=thin_border_side, top=thin_border_side, bottom=thin_border_side)

        # Áp dụng format cho TẤT CẢ các sheets
        for sheet_name in workbook.sheetnames:
            ws = workbook[sheet_name]
            ws.views.sheetView[0].showGridLines = True # Đảm bảo luôn hiện đường grid

            # Format Header (Dòng 1)
            for cell in ws[1]:
                cell.fill = header_fill
                cell.font = header_font
                cell.alignment = Alignment(horizontal="center", vertical="center")

            # Format Data (Từ dòng 2 trở đi)
            for row in ws.iter_rows(min_row=2, max_row=ws.max_row, min_col=1, max_col=ws.max_column):
                for cell in row:
                    cell.font = regular_font
                    cell.border = thin_border

                    # Nếu là sheet Summary và ở dòng cuối cùng (Kết luận) -> In đậm
                    if sheet_name == 'Summary' and cell.row == ws.max_row:
                        cell.font = bold_font

                    # Định dạng Số tiền (Cột chứa chữ 'Amount' hoặc cột số 2 của Summary)
                    col_letter = get_column_letter(cell.column)
                    header_value = ws[f"{col_letter}1"].value

                    if (header_value and 'Amount' in str(header_value)) or (sheet_name == 'Summary' and cell.column == 2):
                        if isinstance(cell.value, (int, float)):
                            cell.number_format = '#,##0;(#,##0);"-"' # Định dạng số kế toán: 1,000,000 hoặc (1,000,000)
                            cell.alignment = Alignment(horizontal="right")
                        else:
                            cell.alignment = Alignment(horizontal="center")

                    # Định dạng Ngày tháng
                    if header_value and 'Date' in str(header_value):
                        cell.number_format = 'yyyy-mm-dd'
                        cell.alignment = Alignment(horizontal="center")

            # Tự động điều chỉnh độ rộng của cột (Auto-fit columns)
            for col in ws.columns:
                max_len = max(len(str(cell.value or '')) for cell in col)
                col_letter = get_column_letter(col[0].column)
                ws.column_dimensions[col_letter].width = max(max_len + 3, 12)

    print(f"🎉 Báo cáo chuẩn FDI đã xuất thành công tại: {output_path}")

# CHẠY THỬ NGHIỆM
bank_reconciliation_fdi('GL_Data.xlsx', 'Bank_Data.xlsx', 'FDI_Bank_Reconciliation_Report.xlsx')
```
